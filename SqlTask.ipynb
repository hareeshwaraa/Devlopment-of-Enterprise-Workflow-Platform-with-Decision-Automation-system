```python
import sqlite3
import pandas as pd

# Create database
conn = sqlite3.connect(":memory:")

# Employee table
conn.execute("""
CREATE TABLE Employee (
    emp_id INTEGER PRIMARY KEY,
    name TEXT,
    department_id INTEGER,
    salary INTEGER,
    city TEXT,
    manager_id INTEGER
)
""")

# Department table
conn.execute("""
CREATE TABLE Department (
    department_id INTEGER PRIMARY KEY,
    department_name TEXT
)
""")

print("Tables created successfully!")
```

```python
employees = [
    (1, 'Ravi', 101, 50000, 'Hyderabad', 5),
    (2, 'Priya', 102, 60000, 'Chennai', 5),
    (3, 'Arun', 101, 45000, 'Hyderabad', 1),
    (4, 'Sita', 103, 70000, 'Bangalore', 5),
    (5, 'Kiran', 101, 90000, 'Hyderabad', None),
    (6, 'Meena', 102, 60000, 'Chennai', 2),
    (7, 'Rahul', 103, 55000, 'Bangalore', 4),
    (8, 'Anu', 104, 40000, 'Mumbai', 1)
]

departments = [
    (101, 'IT'),
    (102, 'HR'),
    (103, 'Sales'),
    (104, 'Finance'),
    (105, 'Marketing')
]

conn.executemany(
    "INSERT INTO Employee VALUES (?, ?, ?, ?, ?, ?)",
    employees
)

conn.executemany(
    "INSERT INTO Department VALUES (?, ?)",
    departments
)

conn.commit()

print("Sample data inserted successfully!")
```

```python
display(pd.read_sql_query("SELECT * FROM Employee", conn))
display(pd.read_sql_query("SELECT * FROM Department", conn))
```

```python
def run_sql(query):
    result = pd.read_sql_query(query, conn)
    display(result)
```

```python
#1. Display all employee records
run_sql("""
SELECT * FROM Employee;
""")
```

```python
#2. Display employee names and salaries
run_sql("""
SELECT name, salary
FROM Employee;
""")
```

```python
#3. Find employees whose salary is greater than 50,000
run_sql("""
SELECT *
FROM Employee
WHERE salary > 50000;
""")
```

```python
#4. Find employees who live in Hyderabad
run_sql("""
SELECT *
FROM Employee
WHERE city = 'Hyderabad';
""")
```

```python
#5. Find employees whose salary is between 40,000 and 60,000
run_sql("""
SELECT *
FROM Employee
WHERE salary BETWEEN 40000 AND 60000;
""")
```
