1. Reverse a number
Question: Write a Python program to reverse a given integer without converting it into a string.
Prompt:Write a Python program to reverse an integer without converting it into a string or using built-in reverse functions. Explain the logic step by step and provide sample input, output, and time and space complexity.

```python
n = int(input("Enter number: "))
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n //= 10

print(rev)
```

2. Check prime number
Question: Write a program to check whether a number is prime without using any built-in prime-checking function.
Prompt:Write a Python program to check whether a given number is prime. Do not use any built-in prime-checking functions. Use loops and conditional statements. Explain the logic, sample input/output, and time and space complexity.

```python
n = int(input("Enter number: "))

if n <= 1:
    print("Not Prime")
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

3. Fibonacci series
Question: Print the first n terms of the Fibonacci series.
Prompt:Write a Python program to generate the first n terms of the Fibonacci series using a for loop without recursion. Explain how the two previous numbers generate the next number. Include sample input/output and complexity.

```python
n = int(input("Enter terms: "))

a, b = 0, 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

4. Factorial of a number
Question: Find the factorial of a number using a loop.
Prompt:Write a Python program to calculate the factorial of a number using a for loop without using the math.factorial() function. Handle negative numbers and explain the logic with sample input/output and complexity.

```python
n = int(input("Enter number: "))
fact = 1

if n < 0:
    print("Not defined")
else:
    for i in range(1, n + 1):
        fact *= i

    print(fact)
```

5. Palindrome string
Question: Check whether a given string is a palindrome without using built-in reverse functions.
Prompt:Write a Python program to check whether a string is a palindrome without using slicing or built-in reverse functions. Use a loop and explain the logic with sample input/output and complexity.

```python
s = input("Enter string: ")

rev = ""

for ch in s:
    rev = ch + rev

if s == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

6. Character frequency count

Question: Count the frequency of each character in a given string.
Prompt:Write a Python program to count the frequency of each character in a string using a dictionary. Do not use the Counter module. Explain the dictionary logic, sample input/output, and time and space complexity.

```python
s = input("Enter string: ")

freq = {}

for ch in s:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print(freq)
```

7. Second-largest element in a list
Question: Find the second-largest distinct element without using sorting.
prompt:Write a Python program to find the second-largest distinct element in a list without using sort(), sorted(), or max(). Handle duplicate values and lists with fewer than two distinct elements. Explain the logic, sample input/output, and complexity.

```python
arr = [10, 20, 40, 30, 40]

first = float('-inf')
second = float('-inf')

for num in arr:
    if num > first:
        second = first
        first = num
    elif first > num > second:
        second = num

if second == float('-inf'):
    print("Second largest does not exist")
else:
    print(second)
```

8. Remove duplicates from a list
Question: Remove duplicate elements while preserving the original order.
Prompt:Write a Python program to remove duplicate elements from a list without using set(). Preserve the original order. Explain the logic, sample input/output, and time and space complexity.

```python
arr = [1, 2, 2, 3, 4, 4]

result = []

for num in arr:
    if num not in result:
        result.append(num)

print(result)
```

9. Anagram strings
Question: Check whether two strings are anagrams without using the Counter module.
Prompt:Write a Python program to check whether two strings are anagrams using dictionaries to count character frequencies. Do not use Counter(). Explain the logic and include sample input/output and complexity.

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

freq1 = {}
freq2 = {}

for ch in s1:
    freq1[ch] = freq1.get(ch, 0) + 1

for ch in s2:
    freq2[ch] = freq2.get(ch, 0) + 1

if freq1 == freq2:
    print("Anagram")
else:
    print("Not Anagram")
```

10. Find missing number in an array
Question: Find the missing number from an array containing numbers from 1 to n.
Prompt:Write a Python program to find one missing number in an array containing distinct numbers from 1 to n. Use the mathematical sum formula without sorting. Explain the logic, sample input/output, and complexity.

```python
arr = [1, 2, 4, 5]
n = 5

total = n * (n + 1) // 2

missing = total - sum(arr)

print("Missing number:", missing)
```
